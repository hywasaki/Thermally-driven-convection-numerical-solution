```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import AutoMinorLocator
from scipy.interpolate import CubicSpline
from scipy.optimize   import brentq

# --- Carga de datos (manteniendo espacios) ---
growth_neg2 = np.loadtxt("growth_rate_Le20Ra-2p0Sc+400                .txt")
Wq_neg2     = np.loadtxt("Wq_Le20Ra-2p0Sc+400                .txt")
Psiq_neg2   = np.loadtxt("Psiq_Le20Ra-2p0Sc+400                .txt")

growth_pos2 = np.loadtxt("growth_rate_Le20Ra+2p0Sc+400                .txt")
Wq_pos2     = np.loadtxt("Wq_Le20Ra+2p0Sc+400                .txt")
Psiq_pos2   = np.loadtxt("Psiq_Le20Ra+2p0Sc+400                .txt")

# --- Definición de q y x ---
nq = growth_neg2.size
q_in = np.arange(1, nq+1) * 0.3 / nq
q_dense = np.linspace(q_in[0], q_in[-1], 10000)

nx = Wq_neg2.shape[0]
x = np.arange(nx) * 0.5

# --- Crear splines y hallar raíces q0 ---
sp_neg2 = CubicSpline(q_in, growth_neg2, bc_type="natural")
sp_pos2 = CubicSpline(q_in, growth_pos2, bc_type="natural")

q0_neg2 = brentq(sp_neg2, q_in[0], q_in[-1])
q0_pos2 = brentq(sp_pos2, q_in[0], q_in[-1])

# --- Encontrar índices más cercanos ---
idx_q0_neg2 = np.argmin(np.abs(q_in - q0_neg2))
idx_q0_pos2 = np.argmin(np.abs(q_in - q0_pos2))

# --- Normalizar modos en q0 ---
omega_neg2_q0 = Wq_neg2[:, idx_q0_neg2] / np.max(np.abs(Wq_neg2[:, idx_q0_neg2]))
psi_neg2_q0   = Psiq_neg2[:, idx_q0_neg2] / np.max(np.abs(Psiq_neg2[:, idx_q0_neg2]))

omega_pos2_q0 = Wq_pos2[:, idx_q0_pos2] / np.max(np.abs(Wq_pos2[:, idx_q0_pos2]))
psi_pos2_q0   = Psiq_pos2[:, idx_q0_pos2] / np.max(np.abs(Psiq_pos2[:, idx_q0_pos2]))

# Texto compacto
info_q0_neg2 = f"q₀={q0_neg2:.3f}"
info_q0_pos2 = f"q₀={q0_pos2:.3f}"

palette = ["#08306B", "#2171B5"]

# --- Gráfica de ω_q en q0 ---
fig_omega0, ax_omega0 = plt.subplots(figsize=(7,4))
ax_omega0.plot(x, omega_neg2_q0, lw=2, color=palette[0], label=f'Ra=-2.0 ({info_q0_neg2})')
ax_omega0.plot(x, omega_pos2_q0, lw=2, color=palette[1], label=f'Ra=+2.0 ({info_q0_pos2})')
ax_omega0.set_xlabel('Distancia')
ax_omega0.set_ylabel(r'Vorticidad $\omega_q$')
ax_omega0.legend(loc='upper left', fontsize=8, title='Rayleigh')
ax_omega0.grid(which='major')
ax_omega0.xaxis.set_minor_locator(AutoMinorLocator())
ax_omega0.yaxis.set_minor_locator(AutoMinorLocator())
ax_omega0.grid(which='minor', linestyle=':', linewidth=0.3, alpha=0.5)
fig_omega0.tight_layout()
fig_omega0.savefig("omegaq_q0_Ra+2_vs_Ra-2.png", dpi=300, bbox_inches="tight")

# --- Gráfica de ψ_q en q0 ---
fig_psi0, ax_psi0 = plt.subplots(figsize=(7,4))
ax_psi0.plot(x, psi_neg2_q0, lw=2, color=palette[0], label=f'Ra=-2.0 ({info_q0_neg2})')
ax_psi0.plot(x, psi_pos2_q0, lw=2, color=palette[1], label=f'Ra=+2.0 ({info_q0_pos2})')
ax_psi0.set_xlabel('Distancia')
ax_psi0.set_ylabel(r'Función de corriente $\Psi_q$')
ax_psi0.legend(loc='upper left', fontsize=8, title='Rayleigh')
ax_psi0.grid(which='major')
ax_psi0.xaxis.set_minor_locator(AutoMinorLocator())
ax_psi0.yaxis.set_minor_locator(AutoMinorLocator())
ax_psi0.grid(which='minor', linestyle=':', linewidth=0.3, alpha=0.5)
fig_psi0.tight_layout()
fig_psi0.savefig("psiq_q0_Ra+2_vs_Ra-2.png", dpi=300, bbox_inches="tight")

plt.show()
```
