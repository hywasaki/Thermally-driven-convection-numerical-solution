# $R_a = 2$

```python
import numpy as np
import matplotlib.pyplot as plt

# --- parámetros y malla 1-D ---
nx   = 2000
nq   = 150
dx   = 0.5
q_max = 0.3
x        = np.arange(nx) * dx
q_vals   = np.arange(1, nq + 1) * (q_max / nq)

# --- carga de datos desde disco ---
C0      = np.loadtxt('C0_save.txt')
T0      = np.loadtxt('T0_save.txt')
Cq      = np.loadtxt('Cq_save.txt')
Tq      = np.loadtxt('Tq_save.txt')
Wq      = np.loadtxt('Wq_save.txt')
Psiq    = np.loadtxt('Psiq_save.txt')
growth  = np.loadtxt('growth_rate.txt')

variables = {
    'C0':   C0,
    'T0':   T0,
    'Cq':   Cq,
    'Tq':   Tq,
    'Wq':   Wq,
    'Psiq': Psiq
}

# --- índices equiespaciados para las variables con gran variación de escala ---
idx_sample = (np.arange(nq) if nq <= 5
              else np.round(np.linspace(0, nq - 1, 5)).astype(int))

# ----------------------------------------------------------------------
for name, mat in variables.items():

    # Elegimos si graficamos todo o solo una muestra
    if name in ['C0', 'T0']:
        idx_to_plot = np.arange(nq)      # todas las columnas
        fig, ax = plt.subplots(figsize=(8, 4.5))
        for idx in idx_to_plot:
            ax.plot(x, mat[:, idx],
                    linewidth=1.0,
                    alpha=0.9)            # curvas superpuestas
        ax.set_xlabel('Distance (x)', fontsize=12)
        ax.set_ylabel(name, fontsize=12)
        ax.set_title(f'{name} vs x  q', fontsize=14)
        ax.grid(True, linestyle='--', alpha=0.6)

    else:                                # Cq, Tq, Wq, Psiq
        for idx in idx_sample:
            fig, ax = plt.subplots(figsize=(8, 4.5))
            ax.plot(x, mat[:, idx],
                    'k--', linewidth=2)
            ax.set_xlabel('Distance (x)', fontsize=12)
            ax.set_ylabel(name, fontsize=12)
            ax.set_title(f'{name} vs x   (q = {q_vals[idx]:.3f})',
                         fontsize=14)
            ax.grid(True, linestyle='--', alpha=0.6)

    plt.tight_layout()
    plt.show()

# --------- gráfica de la tasa de crecimiento σ(q) ----------------------
plt.figure(figsize=(7, 4))
plt.plot(q_vals, growth1, 'k--', linewidth=2,
         marker='o', markerfacecolor='red', markeredgecolor='red')
plt.plot(q_vals, growth, 'k--', linewidth=2,
         marker='o', markerfacecolor='red', markeredgecolor='red')
plt.xlabel('Wave Number q', fontsize=12)
plt.ylabel('Growth Rate σ', fontsize=12)
plt.title('Growth Rate vs q', fontsize=14)
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```
